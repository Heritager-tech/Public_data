```python
import re

def calculate_revenue_5d(df):
    """
    Расчет Revenue 5D для всех лет по формуле:
    <год>, Выручка, тыс. RUB 5D = <год>, Выручка, тыс. RUB * <год>, Общ. Коэфф. Года
    """
    # Найти все года из названий столбцов
    years = set()
    for col in df.columns:
        match = re.search(r'(\d{4}),', col)
        if match:
            years.add(match.group(1))
    
    years = sorted(list(years))
    print(f"Найдены года: {years}\n")
    
    # Для каждого года рассчитать Revenue 5D
    for year in years:
        revenue_col = f"{year}, Выручка, тыс. RUB"
        coeff_col = f"{year},  Общ. Коэфф. Года"
        revenue_5d_col = f"{year}, Выручка, тыс. RUB 5D"
        
        # Проверить наличие необходимых столбцов
        if revenue_col not in df.columns:
            print(f"⚠️ Пропускаю {year}, отсутствует столбец: {revenue_col}")
            continue
        if coeff_col not in df.columns:
            print(f"⚠️ Пропускаю {year}, отсутствует столбец: {coeff_col}")
            continue
        
        # Если столбца Revenue 5D еще нет, создать его
        if revenue_5d_col not in df.columns:
            df[revenue_5d_col] = None
        
        # Рассчитать Revenue 5D для каждой строки
        for idx in df.index:
            revenue = pd.to_numeric(df.loc[idx, revenue_col], errors='coerce')
            coeff = pd.to_numeric(df.loc[idx, coeff_col], errors='coerce')
            
            # Если оба значения есть, рассчитать произведение
            if pd.notna(revenue) and pd.notna(coeff):
                df.loc[idx, revenue_5d_col] = revenue * coeff
            else:
                df.loc[idx, revenue_5d_col] = None
        
        print(f"✓ {year}, Выручка, тыс. RUB 5D рассчитан")
    
    print("\n✓ Расчет Revenue 5D завершен для всех лет!")
    return df

# Выполнить расчет
df = calculate_revenue_5d(df)
```

```python
def classify_company_size_all_years(df):
    """
    Заполняет столбцы "<год>, Размер компании" по значениям
    "<год>, Выручка, тыс. RUB 5D". Значения в столбце — в тысячах.

    Категории (в тыс.):
    - >= 5_000_000 : 'Крупнейшее предприятие'
    - >= 2_000_000 : 'Крупное предприятие'
    - >= 1_000_000 : 'Средне-крупное предприятие'
    - >=   500_000 : 'Среднее предприятие'
    - >=   100_000 : 'Малое предприятие'
    - <    100_000 : 'Микропредприятие'
    """
    years = set()
    for col in df.columns:
        m = re.search(r"(\d{4}),", col)
        if m:
            years.add(m.group(1))
    years = sorted(years)

    thresholds = [
        (5_000_000, 'Крупнейшее предприятие'),
        (2_000_000, 'Крупное предприятие'),
        (1_000_000, 'Средне-крупное предприятие'),
        (500_000, 'Среднее предприятие'),
        (100_000, 'Малое предприятие'),
        (0, 'Микропредприятие'),
    ]

    for year in years:
        rev5d_col = f"{year}, Выручка, тыс. RUB 5D"
        size_col = f"{year}, Размер компании"

        if rev5d_col not in df.columns:
            print(f"⚠️ Пропускаю {year}, отсутствует столбец: {rev5d_col}")
            continue

        # Инициализируем колонку (строго object dtype для смешанных значений)
        df[size_col] = pd.Series([None] * len(df), dtype="object")

        # Конвертация в числовой формат (тысячи)
        vals = pd.to_numeric(df[rev5d_col], errors='coerce')

        # Присваиваем метки
        labels = []
        for v in vals:
            if pd.isna(v):
                labels.append(None)
                continue
            assigned = None
            for thr, label in thresholds:
                if v >= thr:
                    assigned = label
                    break
            labels.append(assigned)

        df[size_col] = labels
        print(f"✓ {year}, Размер компании заполнен")

    print("\n✓ Классификация по размерам для всех годов завершена")
    return df

# Выполнить классификацию
df = classify_company_size_all_years(df)
```

```python
def calculate_dynamics(df):
    """
    Расчет динамики (YoY) для каждого года по формуле:
    <год>, Динамика = (<год>, Выручка, тыс. RUB 5D - <год-1>, Выручка, тыс. RUB 5D) / <год-1>, Выручка, тыс. RUB 5D
    
    Если одно из значений = 0, то пишем "Выручка равна 0"
    """
    # Найти все года из названий столбцов
    years = set()
    for col in df.columns:
        match = re.search(r'(\d{4}),', col)
        if match:
            years.add(int(match.group(1)))
    
    years = sorted(list(years))
    print(f"Найдены года: {years}\n")
    
    # Для каждого года (начиная со второго года) рассчитать динамику
    for i in range(1, len(years)):
        current_year = years[i]
        prev_year = years[i-1]
        
        current_revenue_5d_col = f"{current_year}, Выручка, тыс. RUB 5D"
        prev_revenue_5d_col = f"{prev_year}, Выручка, тыс. RUB 5D"
        dynamics_col = f"{current_year}, Динамика"
        
        # Проверить наличие необходимых столбцов
        if current_revenue_5d_col not in df.columns:
            print(f"⚠️ Пропускаю {current_year}, отсутствует столбец: {current_revenue_5d_col}")
            continue
        if prev_revenue_5d_col not in df.columns:
            print(f"⚠️ Пропускаю {current_year}, отсутствует столбец: {prev_revenue_5d_col}")
            continue
        
        # Если столбца динамики еще нет, создать его
        if dynamics_col not in df.columns:
            df[dynamics_col] = None
        
        # Рассчитать динамику для каждой строки
        for idx in df.index:
            current_val = pd.to_numeric(df.loc[idx, current_revenue_5d_col], errors='coerce')
            prev_val = pd.to_numeric(df.loc[idx, prev_revenue_5d_col], errors='coerce')
            
            # Если одно из значений равно 0 или оба значения отсутствуют
            if (pd.notna(current_val) and current_val == 0) or (pd.notna(prev_val) and prev_val == 0):
                df.loc[idx, dynamics_col] = "Выручка равна 0"
            # Если оба значения есть и не равны нулю, рассчитать динамику
            elif pd.notna(current_val) and pd.notna(prev_val) and prev_val != 0:
                df.loc[idx, dynamics_col] = (current_val - prev_val) / prev_val
            else:
                df.loc[idx, dynamics_col] = None
        
        print(f"✓ {current_year}, Динамика рассчитана")
    
    print("\n✓ Расчет динамики завершен для всех лет!")
    print("⚠️ Примечание: если выручка = 0, показано 'Выручка равна 0'")
    return df

# Выполнить расчет
df = calculate_dynamics(df)
```

```python
def calculate_cagr_with_years(df, min_periods=2):
    """
    Рассчитать CAGR по каждой компании, используя только крайние существующие годы с положительной выручкой (в тыс.).

    Правила:
    - Берём только года, где '<год>, Выручка, тыс. RUB 5D' > 0.
    - Вычисляем первый и последний год с данными: first_year, last_year.
    - periods = last_year - first_year
    - Если periods < min_periods -> не считаем (оставляем пусто)
    - Иначе: CAGR = (last_value / first_value) ** (1/periods) - 1
    - Добавляем два столбца: 'CAGR' (float) и 'CAGR Годы' (строка вида '2015-2024 (9)')
    """
    # Найти доступные годы по шаблону
    years = []
    pattern = re.compile(r"(\d{4}), Выручка, тыс. RUB 5D")
    for col in df.columns:
        m = pattern.search(col)
        if m:
            years.append(int(m.group(1)))
    years = sorted(set(years))
    if not years:
        print("Не найдено столбцов 'Выручка, тыс. RUB 5D'.")
        return df

    # Инициализация столбцов
    df['CAGR'] = pd.Series([None] * len(df), dtype='float64')
    df['CAGR Годы'] = pd.Series([None] * len(df), dtype='object')

    # Для каждой строки определяем первый и последний год с положительной выручкой
    for idx in df.index:
        years_with_values = []
        values = {}
        for y in years:
            col = f"{y}, Выручка, тыс. RUB 5D"
            if col in df.columns:
                val = pd.to_numeric(df.at[idx, col], errors='coerce')
                if pd.notna(val) and val > 0:
                    years_with_values.append(y)
                    values[y] = val

        if len(years_with_values) < 2:
            # недостаточно данных
            continue

        first_year = min(years_with_values)
        last_year = max(years_with_values)
        periods = last_year - first_year

        if periods < min_periods:
            # не считаем, слишком короткий промежуток (например, соседние годы)
            continue

        first_val = values.get(first_year)
        last_val = values.get(last_year)

        if first_val is None or last_val is None or first_val <= 0:
            continue

        try:
            cagr = (last_val / first_val) ** (1.0 / periods) - 1.0
        except Exception:
            cagr = None

        df.at[idx, 'CAGR'] = cagr
        df.at[idx, 'CAGR Годы'] = f"{first_year}-{last_year} ({periods})"

    print(f"✓ CAGR рассчитан (минимум period = {min_periods}) для компаний с достаточными данными")
    return df

# Выполнить расчёт CAGR (минимум 2 промежутка -> период >=2)
df = calculate_cagr_with_years(df, min_periods=2)
```
